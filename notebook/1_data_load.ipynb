# import libraries

```python
import zipfile as zf
import io
import os
import pandas as pd
import numpy as np
```

# read zipfiles

```python
path_zipfile = 'C:\\Users\\z.zarei\\Desktop\\code\\data\\raw_data\\tennis_data.zip'

with zf.ZipFile (path_zipfile,'r') as tennis_data:
    tennis_data= tennis_data.open('20240201.zip').read()
    tennis_data_bytes= io.BytesIO(tennis_data)
    with zf.ZipFile (tennis_data_bytes , 'r') as daily_data:
        daily = daily_data.namelist()
            
parquet =[]
for item in daily:
    if item.endswith('.parquet'):
        parquet.append(item)

print (len(parquet), parquet[0])
```

# creat home and away team score table for one day

```python
table_names = [
    'home_team_score',
    'away_team_score'
]

for table in table_names:
    data = []
    with zf.ZipFile (tennis_data_bytes , 'r') as daily_data:
        for item in parquet:
            filename = os.path.basename(item)
            if filename.startswith(table):
                daily_index = daily_data.open(item).read()
                daily_bytes = io.BytesIO (daily_index)
                df = pd.read_parquet(daily_bytes)
                data.append(df)
    data = pd.concat(data, ignore_index = True)

    folder_path = 'C:\\Users\\z.zarei\\Desktop\\code\\data\\intermedate_data\\20240201'
    os.makedirs(folder_path, exist_ok=True)

    file_path = os.path.join(folder_path, f'{table}.parquet')
    data.to_parquet(file_path, index=False)
    print(f'فایل با موفقیت ذخیره شد: {file_path}')
                
        
```
