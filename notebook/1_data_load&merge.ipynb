# import libraries

```python
import zipfile as zf
import io
import os
import pandas as pd
import numpy as np
```

# concat all tables for 60 days

```python
path_zipfile = 'C:\\Users\\ASMA\\Desktop\\code\\Tennis_project\\data\\raw data\\tennis_data.zip'
output_path = r'C:\Users\ASMA\Desktop\code\Tennis_project\data\intermediate_data'
zip_files =[]
table ={}
final_tables = {}
with zf.ZipFile (path_zipfile,'r') as main_data:
    for file in main_data.namelist():
        if '.zip' in file:
            zip_files.append(file)          
    for zip_file in zip_files:
        daily = main_data.open(zip_file).read()
        daily_bytes = io.BytesIO (daily)
        with zf.ZipFile (daily_bytes , 'r') as daily_data:
            for item in daily_data.namelist():
                if item.endswith('.parquet'):
                    data= daily_data.open(item).read()
                    data_bytes = io.BytesIO (data)
                    df = pd.read_parquet(data_bytes)
                    filename = os.path.basename(item).rsplit('_', 1)
                    if filename[0] not in table:
                        table[filename[0]] = []
                    table[filename[0]].append(df)                  
    for name, dfs in table.items():
        final_tables[name] = pd.concat(dfs, ignore_index=True)
                    
    os.makedirs(output_path, exist_ok=True)
    for name, df in final_tables.items():
        file_path = os.path.join(output_path, name + '.parquet')
        df.to_parquet(file_path, index=False)                   
   
        


                
        
```
